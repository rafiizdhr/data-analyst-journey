# Automatic File Sorter


```python
import os, shutil
```

```python
path = r'D:/DATA ANALYST/data-analyst-journey/PYTHON/automatic file sorter/'
```

```python
file_name = os.listdir(path)
```

```python
#make folders
folder_file = ['CSV Files','JPG Files','Text Files']

for loop in range(0, len(folder_file)):
    if not os.path.exists(path + folder_file[loop]):
        print(path + folder_file[loop])
        os.makedirs(path + folder_file[loop])
```

```python
for file in file_name:
    if '.csv' in file and not os.path.exists(path + "CSV Files/" + file):
        shutil.move(path + file, path + "CSV Files/" + file)
    elif '.jpg' in file and not os.path.exists(path + "JPG Files/" + file):
        shutil.move(path + file, path + "JPG Files/" + file)
    elif '.txt' in file and not os.path.exists(path + "Text Files/" + file):
        shutil.move(path + file, path + "Text Files/" + file)
    else:
        print("there is no file to be move.")
```
