# Visualization with pandas
scatter plot, line plot, bar charts, histograms, and more/


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r'D:\DATA ANALYST\data-analyst-journey\PYTHON\pandas series\file sample\Ice Cream Ratings.csv')
df = df.set_index('Date')
df
```

```python
print(plt.style.available)
plt.style.use('bmh')
```

```python
df.plot(title='Ice Cream Ratings') # or df.plot(kind='line')
```

```python
df.plot(subplots=True, title='Ice Cream Ratings + Subplot')
```

```python
df.plot(title='Ice Cream Ratings', xlabel='Daily Ratings', ylabel='Scores')
```

```python
df.plot(kind='bar')
df.plot(kind='bar', stacked=True)
```

```python
df['Flavor Rating'].plot(kind='bar')
```

```python
df.plot.barh()
df.plot.barh(stacked=True)
```

```python
df.plot.scatter(x = 'Texture Rating', y = 'Flavor Rating', s = 300, c='red')
```

```python
df.plot.hist(bins=10)
df
```

```python
df.boxplot(figsize=(10, 5)) # figsize for bigger output
```

```python
df.plot.area(figsize=(10, 5))
```

```python
df.plot.pie(y='Flavor Rating', figsize=(10, 5))
```
