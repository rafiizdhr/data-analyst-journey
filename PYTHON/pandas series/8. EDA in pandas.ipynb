```python
# 4. want to sorting 2022 population from the most populated country
df.sort_values(by='2022 Population', ascending=False).head(10)
```

```python
# 5. checking correlation between all attribute to do ur analysis
df.corr(numeric_only=True)
```

```python
# 6. better visualizing for correlation
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.rcParams['figure.figsize'] = (20, 10)
plt.show()
```

```python
# 7. eda question : are there any continents that grow faster than other and which ways
df.groupby('Continent').mean(numeric_only=True).sort_values(by='2022 Population', ascending=False)
```
