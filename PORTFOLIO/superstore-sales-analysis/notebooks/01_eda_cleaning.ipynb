### 2. Data Cleaning
- Check duplicates (df.duplicated().sum())
- Check nulls per column
- Fix date columns (Order Date, Ship Date → datetime)
- Standardize category/text casing
- Check for negative or zero Sales/Profit outliers

```python
df.duplicated().sum()
```

```python
df['Postal Code'] = df['Postal Code'].fillna(00000)
```

```python
df.info()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d/%m/%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%d/%m/%Y')
```

### 3. Feature engineering — extract Year, Month, Shipping Duration (Ship Date − Order Date)

```python
df
```

```python
df.info()
```

```python
df.columns
```

```python
df['Order Year'] = df['Order Date'].dt.year
df['Order Month'] = df['Order Date'].dt.month
df['Order Month Name'] = df['Order Date'].dt.month_name()
df['Shipping Duration'] = (df['Ship Date'] - df['Order Date']).dt.days
```

### 4. EDA questions to answer
- Which region/category drives the most sales — and are there big gaps between them?
- Which sub-categories have high total sales but low order volume (or vice versa)?
- Seasonal trend — do sales spike in certain months?
- Top 10 customers by total sales — how concentrated is revenue?

```python
# Which region/category drives the most sales — and are there big gaps between them?
df.groupby('Region')['Sales'].sum(numeric_only=True).sort_values(ascending=False)
```

```python
df.groupby('Category')['Sales'].sum(numeric_only=True).sort_values(ascending=False)
```

```python
# Which sub-categories have high total sales but low order volume (or vice versa)?
df.groupby('Sub-Category')['Sales'].agg(['sum', 'count', 'mean']).sort_values('sum', ascending=False)
```

```python
# Seasonal trend — do sales spike in certain months?
df.groupby('Order Month Name')['Sales'].sum().reindex(['January','February','March','April','May','June',
    'July','August','September','October','November','December'])
```

```python
# Top 10 customers by total sales — how concentrated is revenue?
df.groupby('Customer Name')['Sales'].sum().sort_values(ascending=False).head(10)
```

```python
# Does Ship Mode affect Shipping Duration as expected — and 
df.groupby('Ship Mode')['Shipping Duration'].mean().sort_values()
```

```python
# does faster shipping correlate with order size?
df.groupby('Ship Mode')['Sales'].mean().sort_values(ascending=False)
```
